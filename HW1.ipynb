One thing we did not discuss in class was the use of `group_by` in pandas. `group_by` will partition the data in your data frame by the value of one or more variables - it divides it into chunks. It doesn't create a new data frame - you have to aggregate the values using a second method `agg` to tell it how to combine them. Looking up examples and further explanation of `group_by` and `agg` would be an acceptable and **great** use of AI, but also, you can definitely ask for examples and explanations on Piazza/in office hours.

We will group our data by Park Name and Category and then aggregate by count. There are a few accounting steps to make sure we have a proper index as well. We can do this with this line of code (it's one line, just long):  
  
`species[['Park Name', 'Category', 'Occurrence']].groupby(['Park Name','Category'], as_index = False).agg('count').reset_index(drop = True)`

Use this to transform the data. You will end up with a data frame with 3 columns.

- Brief explanation of what the occurrence column contains: **[Contains the number of animals within a category that are in a specific park.]** 
- Is this dataframe in 3NF? Why or why not? **[Yes. Passes 1NF because df is atomic, order is irrelevant, and primary key is (Park Name, Category). Passes 2NF because there are no partial dependencies as Occurrence value is determined by comination of Park Name & Category. Passes 3NF because there are no transitive depencies (nothing for Occurence to transitively depend on.)]**
- The park with the most bird species (Yo, this is a hint for Blank_1) **[Grand Canyon National Park]**

```python
bla = df[['Park Name', 'Category', 'Scientific Name', 'Occurrence']].groupby(['Park Name','Category', 'Scientific Name'], as_index = False).agg('count').reset_index(drop = True)
bla

bla[bla['Park Name']=='Acadia National Park']
```

```python
# Fill in code here:
new_df = df[['Park Name', 'Category', 'Occurrence']].groupby(['Park Name','Category'], as_index = False).agg('count').reset_index(drop = True)
bird = new_df[new_df['Category']== 'Bird']
bird[bird['Occurrence'] == bird['Occurrence'].max()]

# df
b = df[df['Category'] == 'Bird']
b_spcies = b['Scientific Name'].nunique()
parks = df['Park Name'].nunique()

avg = b_spcies / parks
print(avg)
```

# Question 7 (20 Points)
Create a table that shows the percentage of species of each category of organism in each park (out of the total number of species for just that park). Use the table from the question above and methods we discussed in class - it may be easiest to do this in two steps. Your resulting table will have only 1 row per park and the column names will be the categories (Reptile, Amphibian, etc.). 


- Number of parks with more than 25% of species that are fish **[Blank 1]**
- Category that is on average (just use the mean) the highest percentage across all parks **[Blank 2]**
- Is this table Tidy? Explain briefly why or why not. **[Blank_3]**

```python
# Fill in code here:
# new_df is the df below
df[['Park Name', 'Category','Occurrence']].groupby(['Park Name','Category'], as_index = False).agg('count').reset_index(drop = True)

counts = new_df.pivot_table(
    index='Park Name',     # Column(s) to turn into Row Labels (Groups)
    columns='Category',   # Column(s) to turn into Column Headers
    values='Occurrence',    # Column(s) to aggregate
    aggfunc='sum',      # The math to apply (e.g., 'sum', 'count', 'mean')
).fillna(0)

perc = counts.div(counts.sum(axis=1), axis=0) * 100
perc.head()
```
