## Subquery in SELECT

```python
#What is the highest price for each drink?
```

```python
sql = """
SELECT DISTINCT outer_table.drink_name,
    (
        SELECT MAX(subq_table.price_usd)
        FROM Sells AS subq_table
        WHERE subq_table.drink_name = outer_table.drink_name
    ) AS highest_price
FROM Sells AS outer_table;
"""
```

```python
result = pd.read_sql(sql, conn)
result
```

```python
#I'd probably do this instead
sql = """
SELECT drink_name, MAX(price_usd) AS highest_price
FROM sells
GROUP BY drink_name;
"""
```

```python
pd.read_sql(sql, conn)
```

### Subquery in FROM

```python
# What is the price with tax and a 0.25 surcharge for drinks that end up costing >$6?
```

```python
sql = """
SELECT cafe_name, drink_name, final_price
FROM (
    SELECT cafe_name, drink_name, ROUND(price_usd * 1.08 + 0.25,2) AS final_price
    FROM Sells
) AS sub        
WHERE sub.final_price > 6;
"""

# AS sub is temporary, doesn't exist outside this query
```

```python
result = pd.read_sql(sql, conn)
result
```
