Here is one way to do this:

```python
merged = pd.merge(taxonomy, species_tab, on = 'Scientific Name')
print(merged.shape)
merged2 = pd.merge(merged, parks, on = 'Park Name')
print(merged2.shape)
merged3 = merged2[(merged2['State']== 'VA')&(merged2['Order'] == 'Testudines')]
print(merged3.shape)
#we will get rid of those that were false reports
merged3[~merged3['Occurrence'].str.lower().str.contains('false')]
```
