```python
curs.execute(sql) #different from what we did before - cursor directly executes the statements
```

```python
curs.fetchall() #fetchall will get the results and print them
```

## Create Sells Table

```python
sql ="""
CREATE TABLE Sells(
    cafe_name TEXT NOT NULL,
    drink_name TEXT NOT NULL,
    price_usd REAL NOT NULL CHECK(price_usd > 0),
    PRIMARY KEY(cafe_name, drink_name),
    FOREIGN KEY(drink_name) REFERENCES Drink_Type(drink_name)
);
"""
```

if you run the above code and it works, u know for certain Drink_Type table exists.


```python
curs.execute(sql)
```
