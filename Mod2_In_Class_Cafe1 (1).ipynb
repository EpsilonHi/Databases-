## Subquery in SELECT

```python
#What is the highest price for each drink?
```

```python
sql = """
SELECT DISTINCT outer_table.drink_name,
    (
        SELECT MAX(subq_table.price_usd)
        FROM Sells AS subq_table
        WHERE subq_table.drink_name = outer_table.drink_name
    ) AS highest_price
FROM Sells AS outer_table;
"""
```

```python
result = pd.read_sql(sql, conn)
result
```

### Subquery in FROM

```python
# What is the price with tax and a 0.25 surcharge for drinks that end up costing >$6?
```

```python
sql = """
SELECT cafe_name, drink_name, final_price
FROM (
    SELECT cafe_name, drink_name, ROUND(price_usd * 1.08 + 0.25,2) AS final_price
    FROM Sells
) AS sub
WHERE sub.final_price > 6;
"""
```

```python
result = pd.read_sql(sql, conn)
result
```

### Some set operations

```python
sql = """
(SELECT *
FROM Sells
WHERE price_usd > 5.50)
UNION ALL
(SELECT *
FROM Sells
WHERE price_usd < 3);
"""
#Question...will using just UNION give a different answer?
#Question...what is a better way to do this?
```

```python
result = pd.read_sql(sql, conn)
result
```

```python
sql = """
SELECT *
FROM Sells
WHERE price_usd < 3
UNION ALL
SELECT *
FROM Sells
WHERE cafe_name = 'Starbucks';
"""
#UNION vs UNION ALL
```

```python
result = pd.read_sql(sql, conn)
result
```

```python
sql = """
SELECT *
FROM Sells
WHERE price_usd > 5.50
INTERSECT
SELECT *
FROM Sells
WHERE drink_name = 'Mocha';
"""
#SQLite does not allow the parentheses
```

```python
result = pd.read_sql(sql, conn)
result
```
