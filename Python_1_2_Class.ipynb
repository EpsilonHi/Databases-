```python
def pop_category(population: int)->str:
    '''
    Classify the population into a category based on value
    '''
    if pd.notnull(population):
        if population < 10000000:
            return 'Low'
        elif population < 100000000:
            return 'Medium'
        elif population < 1000000000:
            return 'High'
        else:
            return 'Mega'
        
```

```python
help(pop_category)
```

```python
pop_category.__annotations__
```

```python
cat = df['pop'].apply(pop_category)
df['pop_category'] = cat
df['pop_category'].value_counts()
```
